```python
# linear regression using class
import numpy as np

class myLR:
    def __init__(self):
        self.beta = None
        # placeholder for the model parameters
    
    def fit(self, X, y):
        # Placeholder for the fitting logic
        assert X.shape[0] > 0 #"X must have at least one sample"
        self.beta = np.linalg.pinv(X.T @ X) @ X.T @ y
        # computer the model parameters using the normal equation

    def predict(self, X):
        # Placeholder for the prediction logic
        if self.beta is None:
            raise ValueError("Model is not fitted yet.")

        X_test = np.insert(X,0,1,axis=1)
        return X_test @ self.beta
```

```python
data_x = np.random.randint(0,100,30)
data_y = 2*data_x 
```

```python
# use myLR
model = myLR()
model.fit(data_x.reshape(-1,1), data_y)
```

```python
model.predict(data_x.reshape(-1,1))
```

```python
class animal: # methods and attributes of animal
    def __init__(self, sound, species):
        self.sound = sound
        self.species = species

    def make_sound(self):
        return self.sound

    def get_species(self):
        return self.species

my_Cat = animal("Meow", "Cat")  # Create an instance of the animal class
print(my_Cat.make_sound())  # Output: Meow
print(my_Cat.get_species())  # Output: Cat
```
